# ARO in a notebook

Every cell runs in one session. Variables, feature sets, and user-defined actions
defined in one cell are available in the next.

Run the cells in order.

## Statements and output

`Log` writes to the console as the cell runs. The last statement's result is
displayed automatically — there is no need for an explicit `Return`.

```aro
Log "Starting the tour" to the <console>.
Compute the <total> from 2 + 40.
```

The session remembers `<total>`.

```aro
Compute the <doubled> from <total> * 2.
```

## Collections

A list of records renders as a table — which is most of why data pipelines
(ARO-0018) are worth doing in a notebook.

```aro
Create the <scores> with [
    { name: "Ada", language: "Analytical Engine", score: 99 },
    { name: "Alan", language: "Turing Machine", score: 95 },
    { name: "Grace", language: "COBOL", score: 97 }
].
```

```aro
Reduce the <average: Float> from the <scores> with avg(<score>).
```

```aro
Filter the <top> from the <scores> where <score> > 96.
```

## User-defined actions

A feature set whose activity is `Action` becomes callable as
`Application.<Name>` from any later cell (ARO-0081).

```aro
(Shout: Action takes <text>) {
    Extract the <t> from the <input: text>.
    Compute the <loud: uppercase> from <t>.
    Return an <OK: status> with { loud: <loud> }.
}
```

```aro
Application.Shout the <result> from "notebooks are fun".
Extract the <shouted> from the <result: loud>.
```

## Errors

ARO's philosophy is that the code is the error message (ARO-0006). The cell
below fails on purpose — the error names the statement that caused it.

```aro
Compute the <oops: nosuchqualifier> from <total>.
```
